# Primary setups for the project

## Libraries

```python
from google.cloud import storage
from google.cloud import bigquery
import os
```

### Project setup

```python
# project ID
project = !gcloud config get-value project
PROJECT_ID = project[0]

# region
REGION = 'us-central1'

print(PROJECT_ID)
```

### Adding IM roles for the instance service Account

get the service instance account from workbench, and add the following:  
1 "roles/owner"  
2 "roles/serviceusage.serviceUsageAdmin"  
3 "roles/storage.admin"  
4 "roles/storage.objectAdmin"  

### Clients

```python
gcs = storage.Client(project = PROJECT_ID)
bq = bigquery.Client(project = PROJECT_ID)
```

### Bucket creation within project

```python
BUCKET = PROJECT_ID

if not gcs.lookup_bucket(BUCKET):
    bucketDef = gcs.bucket(BUCKET)
    bucket = gcs.create_bucket(bucketDef, project=PROJECT_ID, location=REGION)
    print(f'Created Bucket: {gcs.lookup_bucket(BUCKET).name}')
    print(f'Review the storage bucket in the console here:\nhttps://console.cloud.google.com/storage/browser/{PROJECT_ID};tab=objects&project={PROJECT_ID}')
else:
    bucketDef = gcs.bucket(BUCKET)
    print(f'Bucket already exist: {bucketDef.name}')
```
